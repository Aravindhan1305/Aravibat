```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("heart.csv")
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

# Scatter plot 1: Age vs Cholesterol
ax1.scatter(df['age'], df['chol'], color='teal', alpha=0.7)
ax1.set_title("Age vs Cholesterol", fontsize=14)
ax1.set_xlabel("Age")
ax1.set_ylabel("Cholesterol")
ax1.grid(True)

# Scatter plot 2: Max Heart Rate vs Resting BP
ax2.scatter(df['thalach'], df['trestbps'], color='coral', alpha=0.7)
ax2.set_title("Max Heart Rate vs Resting BP", fontsize=14)
ax2.set_xlabel("Max Heart Rate")
ax2.set_ylabel("Resting Blood Pressure")
ax2.grid(True)

# Main figure adjustments
fig.suptitle("Heart Data Comparisons Using Matplotlib Anatomy", fontsize=16)
plt.tight_layout()  # Leave space for the super title
plt.show()
```

```python
import matplotlib.pyplot as plt
df_over_50 = df[df["age"] > 50]
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(16, 12), sharex=True)
(ax0, ax1), (ax2, ax3) = axs

# Plot 1: Age vs Cholesterol
sc0 = ax0.scatter(df_over_50["age"], df_over_50["chol"], c=df_over_50["target"], cmap="coolwarm", alpha=0.7)
ax0.axhline(df_over_50["chol"].mean(), color="green", linestyle="--", label="Average")
ax0.set(title="Cholesterol vs Age", ylabel="Cholesterol")
ax0.legend(*sc0.legend_elements(), title="Target")
ax0.grid(True)

# Plot 2: Age vs Max Heart Rate
sc1 = ax1.scatter(df_over_50["age"], df_over_50["thalach"], c=df_over_50["target"], cmap="coolwarm", alpha=0.7)
ax1.axhline(df_over_50["thalach"].mean(), color="green", linestyle="--", label="Average")
ax1.set(title="Max Heart Rate vs Age", ylabel="Max Heart Rate")
ax1.legend(*sc1.legend_elements(), title="Target")
ax1.grid(True)

# Plot 3: Age vs Resting BP
sc2 = ax2.scatter(df_over_50["age"], df_over_50["trestbps"], c=df_over_50["target"], cmap="coolwarm", alpha=0.7)
ax2.axhline(df_over_50["trestbps"].mean(), color="green", linestyle="--", label="Average")
ax2.set(title="Resting BP vs Age", xlabel="Age", ylabel="Resting BP")
ax2.legend(*sc2.legend_elements(), title="Target")
ax2.grid(True)

# Plot 4: Age vs Oldpeak
sc3 = ax3.scatter(df_over_50["age"], df_over_50["oldpeak"], c=df_over_50["target"], cmap="coolwarm", alpha=0.7)
ax3.axhline(df_over_50["oldpeak"].mean(), color="green", linestyle="--", label="Average")
ax3.set(title="Oldpeak vs Age", xlabel="Age", ylabel="Oldpeak")
ax3.legend(*sc3.legend_elements(), title="Target")
ax3.grid(True)

# Main title
fig.suptitle("Heart Disease Feature Comparison (Age > 50)", fontsize=18, fontweight="bold")
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
